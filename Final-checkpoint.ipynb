## Overview

The NHFS was conducted between October 2009 and June 2010. It was one-time survey designed specifically to monitor vaccination during the 2009-2010 flu season in response to the 2009 H1N1 pandemic.

The target population for the NHFS was all persons 6 months or older living in the United States at the time of the interview. Data from the NHFS were used to produce timely estimates of vaccination coverage rates for both the monovalent pH1N1 and trivalent seasonal influenza vaccines.

## Business Understanding

In public health, understanding vaccination patterns is crucial for developing effective campaigns to combat infectious diseases. This project aims to predict whether individuals received the H1N1 and vaccine based on data from the National 2009 H1N1 Flu Survey. The ultimate goal of this project is to create a predictive model that accurately forecasts an individual's vaccination status. This model can serve as a valuable tool for future public health initiatives, allowing for the proactive identification of groups less likely to get vaccinated. Instead of a one-size-fits-all approach, a predictive model enables a tailored response, allocating resources more efficiently to engage and educate hesitant or underserved communities. By transforming survey data into actionable insights, this project supports evidence-based policymaking, ultimately contributing to a more resilient public health system capable of achieving higher rates of herd immunity.

## Model Fine

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
```

```python
# Définir X (features) et y (target)
target_col = 'h1n1_vaccine'
col_drop= ['doctor_recc_seasonal','opinion_seas_vacc_effective','opinion_seas_risk','opinion_seas_sick_from_vacc','hhs_geo_region','h1n1_vaccine','seasonal_vaccine']
y = merged_dff[target_col]
X = merged_dff.drop(col_drop,axis=1) # On retire les deux cibles
```

```python
# Identifications des types de colonnes après le nettoyage initial
# Les variables float64 sont en réalité ordinales (échelle de Likert 0-4 ou binaires 0-1) ou numériques discrètes.

numerical_cols = X.select_dtypes(include=['float64', 'int64']).columns
categorical_cols = X.select_dtypes(include=['object']).columns

# Pipeline pour les colonnes numériques/ordinales
# Remplacons les NaN par la valeur la mode
numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
])

# Pipeline pour les colonnes catégorielles
# 1. Remplacons les NaN par une catégorie 'manquant'
# 2. Appliquons l'encodage One-Hot
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='manquant')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Combiner les transformateurs
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ])
```

```python
# Divisons les données en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42, stratify=y)

# Création d'un pipeline complet 
# Utilisation du modèle de Régression Logistique
model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(solver='liblinear', random_state=42)) # 'liblinear' est bon pour les petits datasets
])

# Entraînement de notre modèle
model.fit(X_train, y_train)

# Évaluation du Modèle
y_pred_proba = model.predict_proba(X_test)[:, 1] # Probabilités pour la classe positive (1)
auc_roc = roc_auc_score(y_test, y_pred_proba)

print(f"\nAUC-ROC du modèle sur l'ensemble de test (h1n1_vaccine): {auc_roc:.4f}")
```

```python
import pandas as pd
import numpy as np

# Récuperation de l'étape de prétraitement (ColumnTransformer)
preprocessor = model.named_steps['preprocessor']

# Récupération des noms de colonnes après le prétraitement

# Noms des colonnes numériques/ordinales (non encodées)
numerical_features = list(X_train.select_dtypes(include=['float64', 'int64']).columns)
# Les noms des colonnes n'ont pas changé dans le premier transformateur (num)

# Noms des colonnes catégorielles après OneHotEncoder
ohe_feature_names = preprocessor.named_transformers_['cat'].named_steps['onehot'].get_feature_names_out(
    X_train.select_dtypes(include=['object']).columns
)

# Combiner tous les noms de colonnes dans l'ordre du pipeline
feature_names = numerical_features + list(ohe_feature_names)

# Récupération des coefficients du modèle de Régression Logistique
# Le classifieur est la dernière étape du pipeline.
coefficients = model.named_steps['classifier'].coef_[0]

# Création du DataFrame pour la visualisation
importance_df = pd.DataFrame({
    'Facteur': feature_names,
    'Coefficient': coefficients
})

# Trie par valeur absolue pour voir les facteurs les plus influents
importance_df['Importance_Absolue'] = np.abs(importance_df['Coefficient'])
importance_df_sorted = importance_df.sort_values(by='Importance_Absolue', ascending=False).drop(columns=['Importance_Absolue'])

# Affichage des facteurs influancant la vaccination
print("Facteurs influençant la probabilité de vaccination (par valeur absolue du coefficient):")
print(importance_df_sorted.head(100))
```
