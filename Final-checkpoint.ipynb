# Flu Shot Learning: Predict H1N1 and Seasonal Flu Vaccines

## Overview

The NHFS was conducted between October 2009 and June 2010. It was one-time survey designed specifically to monitor vaccination during the 2009-2010 flu season in response to the 2009 H1N1 pandemic. The CDC has other ongoing programs for annual phone surveys that continue to monitor seasonal flu vaccination.

The target population for the NHFS was all persons 6 months or older living in the United States at the time of the interview. Data from the NHFS were used to produce timely estimates of vaccination coverage rates for both the monovalent pH1N1 and trivalent seasonal influenza vaccines.

## Business Understanding

In public health, understanding vaccination patterns is crucial for developing effective campaigns to combat infectious diseases. This project aims to predict whether individuals received the H1N1 and seasonal flu vaccines based on data from the National 2009 H1N1 Flu Survey. By analyzing a wide range of factors, including demographics, socioeconomic status, health behaviors, and personal opinions on vaccine effectiveness and illness risk, we can identify key drivers behind vaccination decisions. The insights gained from this analysis will help public health officials and policymakers segment the population and design more targeted, persuasive, and effective communication strategies to encourage higher vaccination rates in future public health crises, similar to the one seen with the H1N1 pandemic. 

The ultimate goal of this project is to create a predictive model that accurately forecasts an individual's vaccination status. This model can serve as a valuable tool for future public health initiatives, allowing for the proactive identification of groups less likely to get vaccinated. Instead of a one-size-fits-all approach, a predictive model enables a tailored response, allocating resources more efficiently to engage and educate hesitant or underserved communities. By transforming survey data into actionable insights, this project supports evidence-based policymaking, ultimately contributing to a more resilient public health system capable of achieving higher rates of herd immunity.

```python
def vaccination_rate_plot(col, target, data, ax=None):
    """Stacked bar chart of vaccination rate for `target` against
    `col`.

    Args:
        col (string): column name of feature variable
        target (string): column name of target variable
        data (pandas DataFrame): dataframe that contains columns
            `col` and `target`
        ax (matplotlib axes object, optional): matplotlib axes
            object to attach plot to
    """
    counts = merged_df[[target, col]].groupby([target, col]).size().unstack(target)
    group_counts = counts.sum(axis="columns")
    props = counts.div(group_counts, axis="index")

    props.plot(kind="barh", stacked=True, ax=ax)
    ax.invert_yaxis()
    ax.legend().remove()
```

```python
cols_to_plot = [
    "h1n1_concern",
    "h1n1_knowledge",
    "opinion_h1n1_vacc_effective",
    "opinion_h1n1_risk",
    "opinion_h1n1_sick_from_vacc",
    "opinion_seas_vacc_effective",
    "opinion_seas_risk",
    "opinion_seas_sick_from_vacc",
    "sex",
    "age_group",
    "race",
]

fig, ax = plt.subplots(len(cols_to_plot), 2, figsize=(9, len(cols_to_plot) * 2.5))
for idx, col in enumerate(cols_to_plot):
    vaccination_rate_plot(col, "h1n1_vaccine", merged_df, ax=ax[idx, 0])
    vaccination_rate_plot(col, "seasonal_vaccine", merged_df, ax=ax[idx, 1])

ax[0, 0].legend(loc="lower center", bbox_to_anchor=(0.5, 1.05), title="h1n1_vaccine")
ax[0, 1].legend(loc="lower center", bbox_to_anchor=(0.5, 1.05), title="seasonal_vaccine")
fig.tight_layout()
```

### Model1

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

from sklearn.linear_model import LogisticRegression
from sklearn.multioutput import MultiOutputClassifier

from sklearn.pipeline import Pipeline

from sklearn.model_selection import train_test_split

from sklearn.metrics import roc_curve, roc_auc_score

RANDOM_SEED = 6 
```

```python
merged_df.dtypes != "object"
```

```python
numeric_cols = merged_df.columns[merged_df.dtypes != "object"].values
```

```python
# chain preprocessing into a Pipeline object
numeric_preprocessing_steps = Pipeline(
    [
        ("standard_scaler", StandardScaler()),
        ("simple_imputer", SimpleImputer(strategy="median")),
    ]
)

# create the preprocessor stage of final pipeline
preprocessor = ColumnTransformer(
    transformers=[("numeric", numeric_preprocessing_steps, numeric_cols)],
    remainder="drop",
)
```

```python
estimators = MultiOutputClassifier(estimator=LogisticRegression(penalty="l2", C=1))
```

```python
ull_pipeline = Pipeline(
    [
        ("preprocessor", preprocessor),
        ("estimators", estimators),
    ]
)
```

```python
y=merged_df[["seasonal_vaccine"]]
X=merged_df
```

```python
X_train, X_eval, y_train, y_eval = train_test_split(X,y,test_size=0.4,random_state=1,
                                                    shuffle=True,stratify=merged_df,#random_state=RANDOM_SEED
)
```

```python
#%%time

# Train model
ull_pipeline.fit(X_train, y_train)

# Predict on evaluation set
preds = ull_pipeline.predict_proba(X_eval)
preds
```

```python
print("test_probas[0].shape", preds[0].shape)
```

```python
y_preds = pd.DataFrame(
    {
        "seasonal_vaccine": preds[0][:, 1],
    },
    index=y_eval.index,
)
print("y_preds.shape:", y_preds.shape)
y_preds.head()
```

```python
from sklearn.metrics import RocCurveDisplay
import matplotlib.pyplot as plt

# Supposons que votre pipeline s'appelle full_pipeline
# et que vous avez déjà X_eval et y_eval

# Trace la courbe ROC pour la cible 'h1n1_vaccine'
RocCurveDisplay.from_estimator(ull_pipeline, X_eval, y_eval['seasonal_vaccine'])

plt.title('Courbe ROC pour H1N1 Vaccine')
plt.xlabel('Taux de faux positifs (FPR)')
plt.ylabel('Taux de vrais positifs (TPR)')
plt.plot([0, 1], [0, 1], 'r--')  # Ligne de référence
plt.show()
```

```python
def plot_roc(y_true, y_score, label_name, ax):
    fpr, tpr, thresholds = roc_curve(y_true, y_score)
    ax.plot(fpr, tpr)
    ax.plot([0, 1], [0, 1], color="grey", linestyle="--")
    ax.set_ylabel("TPR")
    ax.set_xlabel("FPR")
    ax.set_title(f"{label_name}: AUC = {roc_auc_score(y_true, y_score):.4f}")
```

```python
fig, ax = plt.subplots(1, 2, figsize=(7, 3.5))

plot_roc(
    y_eval["seasonal_vaccine"], y_preds["seasonal_vaccine"], "seasonal_vaccine", ax=ax[1]
)
fig.tight_layout()
```
